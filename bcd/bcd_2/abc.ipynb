- **What is Descriptive Statistics?**
    - Descriptive statistics is a branch of statistics that focuses on summarizing and describing the main features of a dataset. It provides simple summaries about the sample and the measures. This can include measures of central tendency (like mean, median, mode) and measures of variability (like standard deviation, variance).


2. Explain the difference between 
    - **mean, median and mode** : 
        - Mean is the average of a set of numbers, calculated by adding all the numbers together and dividing by the count of numbers.
        - Median is the middle value in a sorted list of numbers. If there is an even number of observations, the median is the average of the two middle numbers.
        - Mode is the value that appears most frequently in a dataset. A dataset may have one mode, more than one mode, or no mode at all.
    - **standard deviation , variance , range** :
        - Variance is a measure of how much the values in a dataset differ from the mean and from each other. It is calculated as the average of the squared differences from the mean.
        - Standard deviation is the square root of the variance and provides a measure of the average distance of each data point from the mean. It is expressed in the same units as the data, making it more interpretable than variance.
        - Range is the difference between the maximum and minimum values in a dataset, providing a measure of the spread of the data.

## 3.1 Gaussian Distribution (Normal Distribution)

### Definition
A Gaussian Distribution is a symmetric bell-shaped probability distribution where the mean, median, and mode are equal.

### Formula

$$
f(x)=\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

Where:
- μ = Mean
- σ = Standard Deviation

### Diagram

```text
                 Mean (μ)
                    |
                    |
              .-'''''''-.
           .-'           '-.
         .'                 '.
        /                     \
-------/-----------------------\-------
```

## 3.3 3-Sigma Rule (Empirical Rule)

### Definition
The Empirical Rule states that in a normal distribution:

- 68% of data lies within ±1σ
- 95% of data lies within ±2σ
- 99.7% of data lies within ±3σ

### Formula

$$
\mu \pm 1\sigma
$$

$$
\mu \pm 2\sigma
$$

$$
\mu \pm 3\sigma
$$

### Diagram

```text
          68%       95%        99.7%

----|----|----|----|----|----|----
   -3σ  -2σ  -1σ   μ  +1σ +2σ +3σ
```

## 4. Central Tendency

````python
# calculate mean, median, mode for HouseHoldIncome and Age
mean_income = df['Household_Income'].mean()
median_income = df['Household_Income'].median()
mode_income = df['Household_Income'].mode()[0]  # mode() returns a Series, take the first value

mean_age = df['Age_of_Household_Head'].mean()
median_age = df['Age_of_Household_Head'].median()
mode_age = df['Age_of_Household_Head'].mode()[0]  # mode() returns a Series, take the first value

print(f"Household Income - Mean: {mean_income}, Median: {median_income}, Mode: {mode_income}")
print(f"Age of Household Head - Mean: {mean_age}, Median: {median_age}, Mode: {mode_age}")
````

````python
# interpret the central tendency and comment on income distribution
if mean_income > median_income:
    print("The income distribution is right-skewed, with a long tail on the higher end.")
elif mean_income < median_income:
    print("The income distribution is left-skewed, with a long tail on the lower end.")
````

## 5. Measures of Dispersion

````python
#  Calculate range, variance, standard deviation, and IQR for Household_Income
# Find IQR and comment on income spread


range_income = df["Household_Income"].max() - df["Household_Income"].min()

var_income = df["Household_Income"].var()
std_income = df["Household_Income"].std()

Q1 = df["Household_Income"].quantile(0.25)
Q3 = df["Household_Income"].quantile(0.75)
iqr_income = Q3 - Q1

print("Range:", range_income)
print("Variance:", var_income)
print("Std Dev:", std_income)
print("IQR:", iqr_income)

### Comment
# Large Std Dev → incomes vary widely.  
# IQR shows spread of middle 50% households.
````

## 6. Distribution

````python

import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm
import numpy as np

plt.figure(figsize=(8,5))
sns.histplot(df["Household_Income"], bins=15, kde=False, color="lightblue", stat="density")

mu, sigma = df["Household_Income"].mean(), df["Household_Income"].std()
x = np.linspace(df["Household_Income"].min(), df["Household_Income"].max(), 100)
plt.plot(x, norm.pdf(x, mu, sigma), 'r', linewidth=2)

plt.title("Household Income Distribution with Gaussian Fit")
plt.xlabel("Income")
plt.ylabel("Density")
plt.show()

skewness = df["Household_Income"].skew()
kurtosis = df["Household_Income"].kurt()

print("Skewness:", skewness)
print("Kurtosis:", kurtosis)
````
