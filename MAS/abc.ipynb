```python
import numpy as np
```

```python
data=np.random.randint(0,100,(50))

mu=70
sigma=8
x_bar=np.mean(data)
#  perform one-sample z-test
z_score=(x_bar-mu)/(sigma/np.sqrt(len(data)))
print("Z-score:", z_score)

# p value
from scipy.stats import norm
p_value=2*(1-norm.cdf(abs(z_score)))
print("P-value:", p_value)
```

```python
male_height=np.random.normal(175,10,1000)
female_height=np.random.normal(165,8,1000)

# assume both populations have known standard deviations
sigma_male=10
sigma_female=8

# perform two-sample z-test
x = np.mean(male_height)
y = np.mean(female_height)
z_score_2 = (x - y) / np.sqrt((sigma_male**2 / len(male_height)) + (sigma_female**2 / len(female_height)))
print("Z-score for two-sample test:", z_score_2)

p_value_2 = 2 * (1 - norm.cdf(abs(z_score_2)))
print("P-value for two-sample test:", p_value_2)
```

```python
# generate a sample dataset of 40 student's daily water intake (in liters)
water_intake=np.random.normal(2,0.5,40)

# assume population mean water intake is 2 liters and population standard deviation is 0.4 liters
mu_water=2
sigma_water=0.4

# perform one-sample z-test
x_bar_water=np.mean(water_intake)
z_score_water=(x_bar_water-mu_water)/(sigma_water/np.sqrt(len(water_intake)))
print("Z-score for water intake:", z_score_water)

p_value_water=2*(1-norm.cdf(abs(z_score_water)))
print("P-value for water intake:", p_value_water)
```

```python
# generate 2 datasets of 45 and 50 students scores respectively
scores_class_A=np.random.normal(85,5,45)
scores_class_B=np.random.normal(80,7,50)

# assume both populations have known standard deviations
sigma_A=5
sigma_B=7

# perform two-sample z-test
x_A=np.mean(scores_class_A)
x_B=np.mean(scores_class_B)

z_score_scores=(x_A-x_B)/np.sqrt((sigma_A**2/len(scores_class_A))+(sigma_B**2/len(scores_class_B)))
print("Z-score for scores comparison:", z_score_scores)

p_value_scores=2*(1-norm.cdf(abs(z_score_scores)))
print("P-value for scores comparison:", p_value_scores)
```

```python
# generate a sample dataset of 30 individuals' daily sleep duration (in hours)
sleep_duration=np.random.normal(7,1.5,30)

# assume population mean sleep duration is 8 hours 
population_mean=8

# perform one-sample t-test
from scipy.stats import t
x_bar_sleep=np.mean(sleep_duration)
s = np.std(sleep_duration, ddof=1)  # sample standard deviation
t_score=(x_bar_sleep-population_mean)/(s/np.sqrt(len(sleep_duration)))
print("T-score for sleep duration:", t_score)

# calculate p-value for one-sample t-test
p_value_sleep=2*(1-t.cdf(abs(t_score), df=len(sleep_duration)-1))
print("P-value for sleep duration:", p_value_sleep)
```

```python
# generate 2 independent samples: 40 smokers and 40 non-smokers' daily cigarette consumption
smokers=np.random.normal(15,5,40)   
non_smokers=np.random.normal(0,1,40)

# assume equal and unequal variances as needed
# perform two-sample t-test
from scipy.stats import ttest_ind
t_statistic, p_value_ttest = ttest_ind(smokers, non_smokers, equal_var=False)  # Welch's t-test
print("T-statistic for smokers vs non-smokers:", t_statistic)

print("P-value for smokers vs non-smokers:", p_value_ttest)
```

```python
# generate a sample of 25 teenagers' daily screen time (in hours)
screen_time=np.random.normal(4,1.5,25)

# assume population mean screen time is 4 hours
population_mean_screen_time=4

# perform one-sample t-test
x_bar_screen_time=np.mean(screen_time)
s_screen_time=np.std(screen_time, ddof=1)  # sample standard deviation
t_score_screen_time=(x_bar_screen_time-population_mean_screen_time)/(s_screen_time/np.sqrt(len(screen_time)))
print("T-score for screen time:", t_score_screen_time)

p_value_screen_time=2*(1-t.cdf(abs(t_score_screen_time), df=len(screen_time)-1))
print("P-value for screen time:", p_value_screen_time)
```

```python
# generate 2 samples 35 values for each age group's monthly spending 
age_group_1=np.random.normal(500,100,35)  # Age group 1: 20-30 years
age_group_2=np.random.normal(400,80,35)   # Age group

# assume data are independent and normally distributed
t_statistic_age, p_value_age = ttest_ind(age_group_1, age_group_2, equal_var=False)  # Welch's t-test
print("T-statistic for age groups:", t_statistic_age)
print("P-value for age groups:", p_value_age)
```
