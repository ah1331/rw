```python
import pandas as pd
import numpy as np
```

## step 1: Measures of Central Tendency & Dispersion

```python
df=pd.read_csv(r'C:\Users\ansh\OneDrive\Desktop\Ansh\rw\rw\MAS\final\students_scores.csv')
df.head()
```

```python
# mean , median , mode of math_score

math_mean=df['Math_Score'].mean()
print("Math Score Mean:",math_mean)

math_med=df['Math_Score'].median()
print("Math Score Median:",math_med)

math_mode=df['Math_Score'].mode()[0]
print("Math Score Mode:",math_mode)
```

```python
# find range ,variance , standard deviation of science_score

science_range=df['Science_Score'].max()-df['Science_Score'].min()
print("Science Score Range:",science_range)

science_variance=df['Science_Score'].var()
print("Science Score Variance:",science_variance)

science_std=df['Science_Score'].std()
print("Science Score Standard Deviation:",science_std)
```

## Step 2: Probability Basics

```python
# find the probability of students passing (Pass_Fail=1)
total_students = len(df)
passing_students = len(df[df['Pass_Fail'] == 1])
probability_passing = passing_students / total_students 
print("Probability of students passing:", probability_passing)
```

```python
# create a contingency table between pass_fail and Hours_Studied>5
contingency_table = pd.crosstab(df['Pass_Fail'], df['Hours_Studied'] > 5)
print("Contingency Table:")
display(contingency_table)
```

```python
# calculate conditional probability: P(Pass_Fail=1 | Hours_Studied>5)
conditional_probability = contingency_table.loc[1, True] / contingency_table[True].sum()
print("Conditional Probability (P(Pass_Fail=1 | Hours_Studied>5)):", conditional_probability)
```

## Step 3: Distribution & Visualization

```python
# plot a Histogram & Normal Distribution Curve for Math_Score
import matplotlib.pyplot as plt

plt.hist(df['Math_Score'], bins=20, edgecolor='black')
plt.xlabel('Math Score')
plt.ylabel('Frequency')
plt.title('Distribution of Math Scores')
plt.show()


mean = df['Math_Score'].mean()
std = df['Math_Score'].std()
x = np.linspace(mean - 3*std, mean + 3*std, 100)
plt.plot(x, (1/(std * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean) / std) ** 2))
plt.title('Normal Distribution Curve for Math Scores')
plt.xlabel('Math Score')
plt.ylabel('Probability Density')
plt.show()
```

```python
# calculate Skewness and Kurtosis for Science_Score

skewness = df['Science_Score'].skew()
kurtosis = df['Science_Score'].kurtosis()
print("Skewness of Science Score:", skewness)
print("Kurtosis of Science Score:", kurtosis)
```

```python
# perform a Q-Q plot for English_Score
import scipy.stats as stats

stats.probplot(df['English_Score'], dist="norm", plot=plt)
plt.title('Q-Q Plot for English Scores')
plt.xlabel('Theoretical Quantiles')
plt.ylabel('Sample Quantiles')
plt.show()
```

## Step 4 : Linear Algebra Mini Task

```python
# represent Math_Score and Science_Score of First 5 students as Vectors

vector_math = df['Math_Score'].head(5).values
vector_science = df['Science_Score'].head(5).values

print("Vector for Math Scores of First 5 Students:", vector_math)
print("Vector for Science Scores of First 5 Students:", vector_science)
```

```python
# dot product of the two vectors
dot_product = np.dot(vector_math, vector_science)
print("Dot Product of Math and Science Vectors:", dot_product)
```

```python
# Norm 1 and Norm 2 of the Math_Score vector
norm1 = np.linalg.norm(vector_math, 1)
norm2 = np.linalg.norm(vector_math, 2)
print("Norm 1 of Math Score Vector:", norm1)
print("Norm 2 of Math Score Vector:", norm2)
```

```python
# angle between the two vectors
magnitude_math = np.linalg.norm(vector_math)
magnitude_science = np.linalg.norm(vector_science)

cos_theta = dot_product / (magnitude_math * magnitude_science)
angle_radians = np.arccos(cos_theta)    
degrees = np.degrees(angle_radians)
print("Angle between Math and Science Vectors (in degrees):", degrees)
```
