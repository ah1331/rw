# Part A – Theoretical Foundation 

1. What is inferential statistics?
    - Inferential statistics is a branch of statistics that allows us to make conclusions about a population based on a sample of data. It involves using probability theory to estimate population parameters, test hypotheses, and make predictions.

2. What is hypothesis testing and its components?
    - Hypothesis testing is a statistical method used to make decisions about a population based on sample data. The components of hypothesis testing include:
      - Null Hypothesis (H0): A statement that there is no effect or no difference, which we aim to test against.
      - Alternative Hypothesis (H1): A statement that there is an effect or a difference, which we want to provide evidence for.
      - Test Statistic: A value calculated from the sample data that is used to determine whether to reject the null hypothesis.
      - Significance Level (α): A threshold set by the researcher (commonly 0.05) that determines the probability of rejecting the null hypothesis when it is actually true (Type I error).
      - Decision Rule: A rule that specifies when to reject or fail to reject the null hypothesis based on the test statistic and significance level.

3. Explain confidence interval and critical value.
    - A confidence interval is a range of values, derived from the sample data, that is likely to contain the true population parameter with a certain level of confidence (e.g., 95%). It provides an estimate of the parameter and the uncertainty associated with it.
    - A critical value is a point on the test statistic distribution that corresponds to the significance level (α). It is used to determine the threshold for rejecting the null hypothesis. If the calculated test statistic exceeds the critical value, we reject the null hypothesis.

4. Define p-value.
    - The p-value is the probability of obtaining a test statistic at least as extreme as the one observed, assuming that the null hypothesis is true. It helps to determine the statistical significance of the results. A smaller p-value indicates stronger evidence against the null hypothesis.

5. Differentiate Type I and Type II errors.
    - Type I error (False Positive): This occurs when we reject the null hypothesis when it is actually true. The probability of making a Type I error is denoted by the significance level (α).
    - Type II error (False Negative): This occurs when we fail to reject the null hypothesis when it is actually false. The probability of making a Type II error is denoted by β, and the power of the test (1 - β) represents the probability of correctly rejecting a false null hypothesis.

6. Brief descriptions of z-test, t-test, chi-square test, and ANOVA test.
    - Z-test: A statistical test used to determine whether there is a significant difference between the means of two groups when the population variance is known and the sample size is large.
    - T-test: A statistical test used to compare the means of two groups when the population variance is unknown and the sample size is small. It can be used for independent samples (two-sample t-test) or paired samples (paired t-test).
    - Chi-square test: A statistical test used to determine whether there is a significant association between two categorical variables. It compares the observed frequencies in each category to the expected frequencies under the null hypothesis.
    - ANOVA (Analysis of Variance) test: A statistical test used to compare the means of three or more groups to determine if at least one group mean is significantly different from the others. It analyzes the variance within groups and between groups to make this determination.

7. What is Covariance?
    - Covariance is a measure of how two random variables change together. It indicates the direction of the linear relationship between the variables. A positive covariance means that the variables tend to increase or decrease together, while a negative covariance means that one variable tends to increase when the other decreases. A covariance of zero indicates no linear relationship between the variables.
    
8. What is Correlation?
    - Correlation is a statistical measure that quantifies the strength and direction of the linear relationship between two variables. It is standardized to be between -1 and 1, where a correlation of 1 indicates a perfect positive linear relationship, -1 indicates a perfect negative linear relationship, and 0 indicates no linear relationship. Correlation does not imply causation; it only indicates the degree to which two variables are related.

```python
# 'colums =[record_id', 'age_group', 'age', 'weight', 'gender', 'region',
#    'smoking_status', 'exercise_frequency', 'bmi', 'blood_pressure',
#    'diabetes', 'hypertension', 'cholesterol_level', 'glucose_level',
#    'visit_date']

# 1. Formulate at least two hypotheses from the dataset
# hypothesis 1: There is a significant correlation between smoking status and the prevalence of hypertension among different age groups.
# hypothesis 2: Regular exercise frequency is associated with lower BMI and reduced risk of diabetes in the population.

# 2. Calculate Confidence Intervals for key numerical data like age, weight, etc.
import scipy.stats as stats
import numpy as np
# Example: Confidence Interval for Age
age_mean = df['age'].mean()
age_sem=stats.sem(df['age'])
confidence_level = 0.95
age_ci = stats.t.interval(confidence_level, len(df['age'])-1, loc=age_mean, scale=age_sem)
age_ci_lower, age_ci_upper = age_ci
print(f"Confidence Interval for Age: [{age_ci_lower:.2f}, {age_ci_upper:.2f}]")

# Example: Confidence Interval for Weight
weight_mean = df['weight'].mean()
weight_sem=stats.sem(df['weight'])
weight_ci = stats.t.interval(confidence_level, len(df['weight'])-1, loc=weight_mean, scale=weight_sem)
weight_ci_lower, weight_ci_upper = weight_ci
print(f"Confidence Interval for Weight: [{weight_ci_lower:.2f}, {weight_ci_upper:.2f}]")

```

```python
# 4. t test for mean comparison across groups (e.g., Age between smokers and non-smokers)
from scipy.stats import ttest_ind
smokers_age = df[df['smoking_status'] == 'Smoker']['age']
non_smokers_age = df[df['smoking_status'] == 'Non-Smoker']['age']
t_stat, p_value = ttest_ind(smokers_age, non_smokers_age)

# critical value for 95% confidence level and two-tailed test
critical_value = stats.t.ppf(1 - (1 - confidence_level) / 2, df=len(df)-1)

print(f"T-statistic: {t_stat:.2f}, P-value: {p_value:.4f}")
print(f"Critical Value: {critical_value:.2f}")
if p_value < 0.05:
    print("Reject the null hypothesis: There is a significant difference in age between smokers and non-smokers.")
else:
    print("Fail to reject the null hypothesis: There is no significant difference in age between smokers and non-smokers.")

```

```python
# 5. chi-square test for Smoking habit vs Disease (e.g., Hypertension)
from scipy.stats import chi2_contingency
contingency_table = pd.crosstab(df['smoking_status'], df['hypertension'])
chi2_stat, p_value, dof, expected = chi2_contingency(contingency_table)
critical_value_chi2 = stats.chi2.ppf(1 - (1 - confidence_level), dof)

print(f"Chi-square Statistic: {chi2_stat:.2f}, P-value: {p_value:.4f}")
print(f"Critical Value (Chi-square): {critical_value_chi2:.2f}")
if p_value < 0.05:
    print("Reject the null hypothesis: There is a significant association between smoking status and hypertension.")
else:   
    print("Fail to reject the null hypothesis: There is no significant association between smoking status and hypertension.")
```

```python
# 6. ANOVA test to check if age groups significantly differ in disease rate (e.g., Diabetes)
from scipy.stats import f_oneway
age_groups = df['age_group'].unique()
diabetes_rates = [df[df['age_group'] == group]['diabetes'] for group in age_groups]
f_stat, p_value = f_oneway(*diabetes_rates)
critical_value_anova = stats.f.ppf(1 - (1 - confidence_level), len(age_groups)-1, len(df)-len(age_groups))
print(f"F-statistic: {f_stat:.2f}, P-value: {p_value:.4f}")
print(f"Critical Value (ANOVA): {critical_value_anova:.2f}")
if p_value < 0.05:
    print("Reject the null hypothesis: There is a significant difference in diabetes rates among different age groups.")
else:
    print("Fail to reject the null hypothesis: There is no significant difference in diabetes rates among different age groups.")
    
```
