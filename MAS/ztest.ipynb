**🧪 Question 1: One-Sample z-Test (Two-Tailed)**

📘 Scenario

A smartphone manufacturer claims that the average battery backup of its phone is 24 hours.

A random sample of 36 phones gives:

• Sample mean = 23.4 hours

• Population standard deviation (σ) = 2 hours

Test the claim at α = 0.05.

📝 Tasks
1. State H₀ and H₁
2. Calculate z-score manually
3. Compute p-value
4. Find critical z-value
5. Make a decision using:

    • Method-1 (p-value)

    • Method-2 (critical value)
6. Write final interpretation in plain English


```python
# Hypotheses
# H₀: μ = 1000
# H₁: μ ≠ 1000  (two-tailed)

import numpy as np
from scipy.stats import norm

x_bar = 23.4
mu = 24
sigma = 2
n = 36
alpha = 0.05

z_score = (x_bar - mu) / (sigma / np.sqrt(n))
p_value = 2 * (1 - norm.cdf(abs(z_score)))

critical_z = norm.ppf(1 - alpha/2)

z_score, p_value, critical_z
```

```python
# Method-1: p-value approach

if p_value < alpha:
    print("The p-value is less than 0.05, so we reject the null hypothesis. The average bottle volume is significantly different from 1000 ml.")
else:
    print("The p-value is greater than or equal to 0.05, so we fail to reject the null hypothesis. The average bottle volume is not significantly different from 1000 ml.")
```

```python
# Method-2: Critical z-value approach

if (z_score > critical_z) or (z_score < -critical_z):
    print("The z-score falls outside the decision boundary, so we reject the null hypothesis. The average bottle volume is significantly different from 1000 ml.")
else:
    print("The z-score falls within the decision boundary, so we fail to reject the null hypothesis. The average bottle volume is not significantly different from 1000 ml.")
```

**🧪 Question 2: One-Sample z-Test (Right-Tailed)**

📘 Scenario

An online learning platform claims that the average daily study time of users is at most 4 hours.

A sample of 49 users shows:

• Sample mean = 4.3 hours

• σ = 0.7 hours

Test the claim at α = 0.05.

📝 Tasks
1. Frame right-tailed hypotheses
2. Perform manual z-test
3. Calculate one-tailed p-value
4. Identify critical z
5. Apply both decision methods
6. Interpret result as a business insight

```python
# Hypotheses
# H₀: μ ≤ 3
# H₁: μ > 3  (right-tailed)

x_bar = 4.3
mu = 4
sigma = 0.7
n = 49
alpha = 0.05

z_score = (x_bar - mu) / (sigma / np.sqrt(n))
p_value = 1 - norm.cdf(z_score)

critical_z = norm.ppf(1 - alpha)

z_score, p_value, critical_z
```

```python
# Method-1: p-value approach

if p_value < alpha:
    print("The p-value is less than 0.05, so we reject the null hypothesis. The average study time has significantly increased.")
else:
    print("The p-value is greater than or equal to 0.05, so we fail to reject the null hypothesis. There is no significant increase in study time.")
```

```python
# Method-2: Critical z-value approach

if z_score > critical_z:
    print("The z-score falls outside the decision boundary, so we reject the null hypothesis. The average study time has significantly increased.")
else:
    print("The z-score falls within the decision boundary, so we fail to reject the null hypothesis. There is no significant increase in study time.")
```

**🧪 Question 3: Two-Sample z-Test (Two-Tailed)**

📘 Scenario

Two logistics companies claim to have the same average delivery time.

Company A:
Mean = 42 min, Std = 6 min, n = 64

Company B:
Mean = 39 min, Std = 6 min, n = 81

At α = 0.05, test whether there is a significant difference in delivery times.

📝 Tasks
1. Define H₀ and H₁
2. Compute two-sample z-score
3. Calculate two-tailed p-value
4. Determine critical z
5. Decide using:
• Method-1 (p-value)
• Method-2 (critical boundary)
6. Conclude whether companies differ significantly

```python
# Hypotheses
# H₀: μ₁ = μ₂
# H₁: μ₁ ≠ μ₂  (two-tailed)

x1, x2 = 42, 39
sigma1, sigma2 = 6,6
n1, n2 = 64,81
alpha = 0.05

z_score = (x1 - x2) / np.sqrt((sigma1**2/n1) + (sigma2**2/n2))
p_value = 2 * (1 - norm.cdf(abs(z_score)))

critical_z = norm.ppf(1 - alpha/2)

z_score, p_value, critical_z
```

```python
# Method-1: p-value approach

if p_value < alpha:
    print("The p-value is less than 0.05, so we reject the null hypothesis. The delivery times of the two companies are significantly different.")
else:
    print("The p-value is greater than or equal to 0.05, so we fail to reject the null hypothesis. There is no significant difference in delivery times.")
```

```python
# Method-2: Critical z-value approach

if (z_score > critical_z) or (z_score < -critical_z):
    print("The z-score falls outside the decision boundary, so we reject the null hypothesis. The delivery times are significantly different.")
else:
    print("The z-score falls within the decision boundary, so we fail to reject the null hypothesis. The delivery times are not significantly different.")
```

**🧪 Question 4: Two-Sample z-Test (Left-Tailed)**

📘 Scenario

A new software algorithm claims to reduce system response time compared to the old algorithm.

Algorithm Old:
Mean = 8.5 sec, Std = 1.5 sec, n = 100

Algorithm New:
Mean = 7.9 sec, Std = 1.5 sec, n = 90

Test the claim at α = 0.05.

📝 Tasks
1. Write left-tailed hypotheses
2. Perform manual two-sample z-test
3. Compute one-tailed p-value
4. Identify critical z
5. Apply both decision methods
6. State conclusion in technical + non-technical language

```python
# Hypotheses
# H₀: μ_new ≥ μ_old
# H₁: μ_new < μ_old  (left-tailed)

x_old, x_new = 8.5, 7.9
sigma_old, sigma_new = 1.5,1.5
n_old, n_new = 100,90
alpha = 0.05

z_score = (x_new - x_old) / np.sqrt((sigma_new**2/n_new) + (sigma_old**2/n_old))
p_value = norm.cdf(z_score)

critical_z = norm.ppf(alpha)

z_score, p_value, critical_z
```

```python
# Method-1: p-value approach

if p_value < alpha:
    print("The p-value is less than 0.05, so we reject the null hypothesis. The new algorithm is significantly faster.")
else:
    print("The p-value is greater than or equal to 0.05, so we fail to reject the null hypothesis. The new algorithm is not significantly faster.")
```

```python
# Method-2: Critical z-value approach

if z_score < critical_z:
    print("The z-score falls outside the decision boundary, so we reject the null hypothesis. The new algorithm is significantly faster.")
else:
    print("The z-score falls within the decision boundary, so we fail to reject the null hypothesis. The new algorithm is not significantly faster.")
```
